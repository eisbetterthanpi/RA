````python
# @title function calling tut
# https://colab.research.google.com/github/google-gemini/cookbook/blob/main/quickstarts/Function_calling.ipynb#scrollTo=rjkZ8MA00Coc

def power_disco_ball(power: bool) -> bool:
    """Powers the spinning disco ball."""
    print(f"Disco ball is {'spinning!' if power else 'stopped.'}")
    return True

def start_music(energetic: bool, loud: bool, bpm: int) -> str:
    """Play some music matching the specified parameters.

    Args:
      energetic: Whether the music is energetic or not.
      loud: Whether the music is loud or not.
      bpm: The beats per minute of the music.

    Returns: The name of the song being played.
    """
    print(f"Starting music! {energetic=} {loud=}, {bpm=}")
    return "Never gonna give you up."

def dim_lights(brightness: float) -> bool:
    """Dim the lights.

    Args:
      brightness: The brightness of the lights, 0.0 is off, 1.0 is full.
    """
    print(f"Lights are now set to {brightness:.0%}")
    return True

tools = [power_disco_ball, start_music, dim_lights]
model = genai.GenerativeModel(model_name="gemini-1.5-flash", tools=tools)

chat = model.start_chat()
response = chat.send_message("Turn this place into a party!")


for part in response.parts:
    if fn := part.function_call:
        args = ", ".join(f"{key}={val}" for key, val in fn.args.items())
        # args = ", ".join(f"{key}={val}" for key, val in sorted(fn.args.items()))
        print(f"{fn.name}({args})")
        fn.name(**args)

function_call = response.candidates[0].content.parts[0].function_call


responses = {"power_disco_ball": True, "start_music": "Never gonna give you up.", "dim_lights": True,} # Simulate the responses from the specified tools.
response_parts = [genai.protos.Part(function_response=genai.protos.FunctionResponse(name=fn, response={"result": val})) for fn, val in responses.items()]

response = chat.send_message(response_parts)
print(response.text)
print(chat.history)
````

````python
# @title function calling

# cloud.google.com/vertex-ai/generative-ai/docs/multimodal/function-calling
# ai.google.dev/gemini-api/tutorials/extract_structured_data

semantic_prompt = '''Use the update_semantic_memory function to update the wiki pages of relevant entities based on the input text. The input text may contain information about people, places, activities, things etc.
Call the function once for each entity that needs updating. multiple updates to the same entity should be separated by a new line.
The text should be based on the information in the input text only, do not use your own knowledge
The text formatted in a way that is suitable for the wiki style.

Example:
For the input text "sally: yeah, I recently got a job at gibbs. ty is there too.", the output should be the following update calls:
update_semantic_memory('Sally', 'Sally is working at Gibbs around {date}.\nSally is working at Gibbs with coworker ty.\n')
update_semantic_memory('Ty', 'Ty is working at Gibbs.')
update_semantic_memory('Gibbs', 'Sally and Ty are working at Gibbs around {date}.')'''


def update_semantic_memory(article: str, text: str) -> str:
    """update semantic memory
    update the text of the article

    Args:
      article: The name of the article to update.
      text: The text of the article to update.
    """
    # return
    return article+': '+text


import google.generativeai as genai
from google.colab import userdata
API_KEY = userdata.get('google_ai')
genai.configure(api_key=API_KEY)

# from vertexai.generative_models import FunctionDeclaration, Tool
# tools = Tool(function_declarations=[update_semantic_memory,],)

# generation_config = {"temperature": 0.9, "top_p": 1, "top_k": 1, "max_output_tokens": 2048,} # https://ai.google.dev/api/python/google/generativeai/GenerationConfig
generation_config = {"temperature": 1.9, "top_p": 0.1, "top_k": 100, "max_output_tokens": 2048} # https://cloud.google.com/vertex-ai/generative-ai/docs/learn/prompts/adjust-parameter-values#whats_next
model = genai.GenerativeModel(model_name="gemini-1.5-flash", generation_config=generation_config, system_instruction=semantic_prompt, tools=[update_semantic_memory]) # https://ai.google.dev/api/python/google/generativeai/GenerativeModel
# model = genai.GenerativeModel(model_name="gemini-1.5-flash", generation_config=generation_config, system_instruction=semantic_prompt, tools=tools) # https://ai.google.dev/api/python/google/generativeai/GenerativeModel
# model = genai.GenerativeModel(model_name="gemini-1.5-flash", tools=[update_semantic_memory])

convo = model.start_chat() # Manual function calling


hist = '''With a playful jab, I help the person who matters most.'''
response = convo.send_message(hist)


# summary = response.candidates[0].content.parts[0].text
# print(response.candidates[0].content.parts)

# if part.function_call:
#     result = update_semantic_memory(part.function_call, functions)

# print(result)
````
